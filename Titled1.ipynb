# Monthly Sales Analysis
This notebook generates sample sales data and plots total sales per product.

```python
import pandas as pd

# Create sample sales data
data = {
    'Product': ['Laptops', 'Smartphones', 'Tablets', 'Headphones', 'Monitors'],
    'Units_Sold': [120, 340, 180, 290, 95],
    'Price_Per_Unit': [1000, 700, 400, 150, 300]
}

# Load into a DataFrame
df = pd.DataFrame(data)

# Calculate total revenue
df['Total_Revenue'] = df['Units_Sold'] * df['Price_Per_Unit']

# Display the table
df
```

```python
import matplotlib.pyplot as plt

# Plot a bar chart of Total Revenue by Product
plt.figure(figsize=(8, 4))
plt.bar(df['Product'], df['Total_Revenue'], color='yellow')
plt.title('Total Revenue by Product')
plt.xlabel('Product')
plt.ylabel('Revenue ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Display the chart inline
plt.show()
```
