# Annual Country-Level Loan Disbursement Analysis-2025

### Overview
This notebook analyzes clients and Total Loans Disbursed (in USD) by Country in 2025.

### Key Outputs
* **Summary Dataset:** Country-level client counts and aggregate loan disbursement totals.
* **Metric Calculation:** Average loan size disbursed per client by country.
* **Data Visualization:** Comparative bar chart highlighting total capital deployed across active regions.

```python
import pandas as pd
# Amount in USD
data = {
    'Country': ['Kenya', 'Peru', 'Costa Rica', 'Ghana', 'Uganda'],
    'Clients_Served': [30, 50, 15, 62, 25],
    'Total_Loans_Disbursed_USD': [1500000, 950000, 800000, 600000, 1100000]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate Average Loan Size per Client
df['Avg_Loan_Per_Client'] = df['Total_Loans_Disbursed_USD'] / df['Clients_Served']

# Display the table
df
```

```python
import matplotlib.pyplot as plt

# Plot Total Loans Disbursed by Country
plt.figure(figsize=(9, 4.5))
plt.bar(df['Country'], df['Total_Loans_Disbursed_USD'] / 1e6, color='Purple')

plt.title('Total Loans Disbursed by Country in 2025')
plt.xlabel('Country')
plt.ylabel('Loans Disbursed ')
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Display the chart inline
plt.show()
```
